```python
## Multiplication of array

import numpy as np

array1 = np.array([[1,2],[3,4]])
array2 = np.array ([[5,6],[7,8]])


mul = np.dot(array1, array2)

print (mul)
```

```python
## Mean value

data = np. array ([1,2,3,4,5])

np.mean(data)
```

```python
##STD value

sd = np. array ([1,2,3,4,5])

np.std(sd)
```

```python
##Corrilation

data2 = np. array ([11,12,13,14,15])

cor = np.corrcoef(data,data2)

print(cor)
```
