```python
#By using this we can extract the information from the webpage


import requests # this import request http ask for request
from bs4 import BeautifulSoup

url = 'https://www.goal.com/en-us/lists/usmnt-midfielder-europe-as-weston-mckennie-juventus-americans-abroad/bltcb1aa5c0d3976f26'
response = requests. get(url)

html_content = response.content

print(html_content)
```

```python
soup = BeautifulSoup(html_content, 'html.parser')
h1_tags = soup.find('h1')
print(h1_tags.text)
```

```python
imgs = soup.find_all('img')

for img in imgs:
  print(img['src'])
```

```python
mylist = soup.find('ul')

print(mylist.text)
```

```python
text = mylist.text

with open ('web-script.txt', 'w') as file:
  file.write(text)
```
