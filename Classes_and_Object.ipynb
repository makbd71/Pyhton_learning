```python
class Employee:
  def __init__(self,name,position, salary):
    self.name = name
    self.position = position
    self.salary = salary

  def display_info(self): #def display_info(self):#
      return f"{self.name} is a {self.position} earning ${self.salary} annually"
```

```python
employee1 = Employee("John Doe", "Manager", 75000)
employee2 = Employee("John Smith", "Manager", 65000)
```

```python
print(employee1.display_info())
print(employee2.display_info())
```

```python
employee1.salary = 100000
print(employee1.display_info())
```

```python
class BankAccount:
    def __init__(self, account_id, balance=1000):
        self.account_id = account_id
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
        else:
            print("Invalid deposit amount")

    def get_balance(self):
        return self.balance


def display_balance(amount):
    return f"Your balance is ${amount}"


account1 = BankAccount("JJH100009876", 100)

# Get balance and display it
balance = account1.get_balance()
print(display_balance(balance))
```
