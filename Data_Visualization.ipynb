```python
import pandas as pd

import matplotlib.pyplot as plt
```

```python
data = {'Date': [ '2023-08-01', '2023-08-01','2023-08-01'],
        'Price': [100,105,110]}

df = pd.DataFrame(data)

print(df)
```

```python
df.plot(x= 'Date', y = 'Price')
plt.xlabel('Date')
plt.ylabel('Price')
```

```python
data_product_sales = {'Product': ['Milk', 'Suger', 'Oil'],
                      'Price': [10, 5, 8]
    }

df = pd.DataFrame (data_product_sales)

print(df)
```

```python
df.plot (x='Product', y= 'Price', kind = 'bar')
```

```python
exam_data = {'Hours_stuided': [2,3,4,5],
           'Exam_Score': [60,70,80,90]}
df = pd.DataFrame(exam_data)

print(df)
```

```python
df.plot(x='Hours_stuided', y= 'Exam_Score', kind = 'scatter')
```

```python
business_student = { 'Dept':['Markerting', 'SCM', 'Finance'],
                    'Student_num': [25,23,18]}
df = pd.DataFrame(business_student)
print(df)
```

```python
df.plot(x= 'Dept', y='Student_num', kind = 'pie')
```
